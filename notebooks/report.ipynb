```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
from src.citymove.dataio import load_network
from src.citymove.analytics import route_headways, busiest_stops, plot_headway_histogram, plot_busiest_stops

FEED = ROOT / "data" / "gtfs_static"
network = load_network(FEED)
calendar = pd.read_csv(FEED / "calendar.txt", dtype={"service_id": str})
weekday_services = set(calendar.loc[calendar["monday"] == 1, "service_id"])  # лише будні рейси
```

## Графік 1. Розподіл інтервалів руху (будні)

```python
plot_headway_histogram(route_headways(list(network.trips.values()), service_ids=weekday_services))
plt.show()
```

## Графік 2. Top-10 найзавантаженіших зупинок за кількістю з'єднань

```python
plot_busiest_stops(busiest_stops(network, top=10))
plt.show()
```
